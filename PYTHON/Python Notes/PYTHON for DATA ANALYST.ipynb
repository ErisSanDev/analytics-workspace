***Section 6: Operators***

**An operator is a symbol that performs a specific operation between two operands**. Together with values, operators are combined to form **expressions**, which are the most basic kind of programming instructions.

---

### **Categories of Operators**

**1. Arithmetic Operators**
These are used to perform mathematical calculations between two numerical operands. They include:
*   **Addition (`+`):** Adds values together.
*   **Subtraction (`-`):** Subtracts the right value from the left value.
*   **Multiplication (`*`):** Multiplies values.
*   **Division (`/`):** Divides the left value by the right value (always returning a floating-point decimal).
*   **Integer Division / Floor Division (`//`):** Divides the values and rounds the result down to the nearest whole integer, discarding the remainder.
*   **Modulus / Remainder (`%`):** Divides the values and returns only the remainder of the division.
*   **Exponent (`**`):** Raises the left value to the power of the right value.

**2. String Operators in Python**

i. **String Concatenation (`+`)**
   When the `+` operator is placed between two string values, it acts as the **concatenation operator**. It joins the two strings together to produce a single, brand-new string. 
   * *Example:* `'Alice' + 'Bob'` evaluates to `'AliceBob'`.
   * *Note:* You cannot concatenate a string with an integer directly (e.g., `'Alice' + 42`); this will raise a `TypeError`. You must first explicitly convert the integer using the `str()` function.

ii. **String Replication (`*`)**
   When the `*` operator is used with one string value and one integer value, it becomes the **string replication operator**. It repeats the original string a number of times equal to the integer.
   * *Example:* `'Alice' * 5` evaluates to `'AliceAliceAliceAliceAlice'`.
   * *Note:* You can only use string replication with a string and an integer; multiplying two strings (e.g., `'Alice' * 'Bob'`) or a string and a float (e.g., `'Alice' * 5.0`) will result in a `TypeError`.

iii. **String Membership Operators (`in` and `not in`)**
   These operators can be used with strings to check if a specific substring exists inside another string. They always evaluate down to a Boolean `True` or `False`.
   * *Example:* `'Hello' in 'Hello, World'` evaluates to `True`, while `'cats' not in 'cats and dogs'` evaluates to `False`. These comparisons are strictly case-sensitive.

**3. Comparison (Relational) Operators**
These operators compare two values and always evaluate down to a single Boolean value (**`True`** or **`False`**). They are commonly used to compare a variable's value against another value to make decisions. They include:
*   **`==`** (Equal to) and **`!=`** (Not equal to) — these can be used with values of any data type.
*   **`>`** (Greater than), **`<`** (Less than), **`>=`** (Greater than or equal to), and **`<=`** (Less than or equal to) — these work properly only with numbers.

**4. Logical (Boolean) Operators**
Used to compare Boolean values and evaluate multiple expressions to make complex decisions. They are:
*   **`and`:** Returns `True` only if both expressions on either side are true.
*   **`or`:** Returns `True` if at least one of the expressions on either side is true.
*   **`not`:** Reverses the Boolean value of the expression that follows it (evaluates to `True` if the expression is false, and vice versa).
  
**5. Assignment Operators**
These are used to assign the value of a right-hand expression to a left-hand variable (or operand). 
*   **Basic Assignment (`=`):** Stores a value directly (e.g., `spam = 42`).
*   **Augmented Assignment (`+=`, `-=`, `*=`, `/=`, `**=`, `//=`, etc.):** These operators provide a shortcut to perform a mathematical operation and an assignment in a single step. For instance, `a += 3` is the equivalent of writing `a = a + 3`.

**6. Membership Operators**
These operators are used to verify whether a specific value exists inside a sequence or collection data structure, such as a list or string. They return a Boolean value:
*   **`in`:** Evaluates to `True` if the value is found inside the collection.
*   **`not in`:** Evaluates to `True` if the value is absent from the collection.

**7. Bitwise Operators in Python**
Bitwise operators perform operations directly on the **binary representations (bits)** of integers.
* Suppose we have two variables:
*   `a = 10` (which is `1010` in binary)
*   `b = 4`  (which is `0100` in binary)

The standard bitwise operators are as follows:

i. **Bitwise AND (`&`)**
   Compares each bit of the first operand with the corresponding bit of the second operand. If both bits are `1`, the resulting bit is set to `1`; otherwise, it is set to `0`.
   * *Example:* `a & b` (comparing `1010` and `0100`) evaluates to `0` (binary `0000`).

ii. **Bitwise OR (`|`)**
   Compares each bit of the first operand with the corresponding bit of the second operand. If at least one of the bits is `1`, the resulting bit is set to `1`.
   * *Example:* `a | b` (comparing `1010` and `0100`) evaluates to `14` (binary `1110`).

iii. **Bitwise XOR (`^`)**
   Compares each bit of the first operand with the corresponding bit of the second operand. The resulting bit is set to `1` if only one of the bits is `1` (but not both).
   * *Example:* `a ^ b` evaluates to `14` (binary `1110`).

iv. **Bitwise NOT / One's Complement (`~`)**
   Inverts all the bits of the operand, turning `1`s into `0`s and `0`s into `1`s.
   * *Example:* `~a` evaluates to `-11`.

v. **Left Shift (`<<`)**
   Shifts the binary digits of the first operand to the left by the number of positions specified by the second operand, filling the vacant spaces on the right with `0`s.
   * *Example:* `a << 1` shifts `1010` left by one position to become `20` (binary `10100`).

vi. **Right Shift (`>>`)**
   Shifts the binary digits of the first operand to the right by the number of positions specified by the second operand.
   * *Example:* `a >> 1` shifts `1010` right by one position to become `5` (binary `0101`).

**8. Identity Operators**
**Identity operators in Python** are used to check whether two variables refer to the **same object in memory**.

Python has two identity operators:

* **`is`** – Returns `True` if both variables refer to the same object.
* **`is not`** – Returns `True` if both variables refer to different objects.

**TYPECASTING / TYPE CONVERSION**

***Type Casting***

Typecasting (also referred to as type conversion) is the process of explicitly converting a value from one data type into another

* Because Python is dynamically typed and binds values implicitly to their types, it does not automatically convert mismatched types for certain operations
* For instance, attempting to concatenate a string with an integer (such as 'Alice' + 42) results in a TypeError because Python cannot automatically combine these distinct data types
* To resolve this and prevent your program from crashing, you must explicitly cast the value to the appropriate data type using Python's built-in conversion functions

**Primary Typecasting Functions**

Your sources outline several essential functions used to convert between basic and collection data types:

**str():** Converts a value into its string (text) representation
- This is highly useful when you have an integer or float that you need to concatenate with another string
- For example, casting the float 50.75 into the string '50.75'

**int():** Converts a value into its integer (whole number) form
- This is particularly helpful when converting text-based numbers (such as string digits returned by the input() function) into actual numbers so that you can perform mathematical calculations
- Additionally, passing a floating-point number to int() will round it down to the nearest whole integer (for example, int(7.7) evaluates to 7)

**float():** Converts a value into its floating-point (decimal) representation
- For instance, casting the integer 25 into the float 25.0

**list():** Converts a sequence value, such as a tuple or a string, into a list
- This is handy when you need to create a mutable (changeable) version of an unchangeable tuple

**tuple():** Converts a sequence value, like a list or string, into an unchangeable (immutable) tuple.

**Section 10 - Loops**

***LOOPS***

**What are Loops**

**Loops allow a program to execute a statement or group of statements multiple times**. Instead of executing lines of code strictly from top to bottom, loops allow your programme to repeat instructions based on whether specific conditions are met. This repetition is what enables computers to automate massive, repetitive tasks in a matter of seconds.

* Python features two primary types of loops: **`for` loops** and **`while` loops**.

---
**1. For Loops**
A **`for` loop** is designed to iterate over a sequence of values, such as a list, tuple, dictionary, set, or string. It executes the loop block once for each item present in the sequence.

*   **The `range()` Function:** This built-in function is frequently used with `for` loops to generate a sequence of numbers. By default, it starts at `0` and increments by `1`, ending at (but not including) the number you pass to it. 
    *   You can customise `range()` by passing starting, stopping, and stepping arguments (e.g., `range(start, stop, step)`). Passing a negative integer as the step argument allows the loop to count down instead of up.
*   **The `enumerate()` Function:** When iterating over a list, instead of using `range(len(someList))` to track the index numbers, you can call `enumerate()`. On each iteration, it returns both the index of the item and the item itself.


**2. While Loops**
A **`while` loop** repeats a block of code (the loop body or clause) repeatedly for as long as a specified condition remains `True`. 

*   **Condition Checking:** The interpreter evaluates the condition *before* executing the loop body. When the execution reaches the end of the loop's clause, it jumps back to the start of the `while` statement to re-evaluate the condition.
*   **Infinite Loops:** If the loop's condition never becomes `False`, the programme will be trapped in an **infinite loop**. If your programme is stuck in an infinite loop, you can stop it manually by pressing **`CTRL-C`** in your terminal.
*   **While with Else:** You can optionally combine a `while` loop with an **`else` statement**. The code inside the `else` block runs exactly once when the loop condition evaluates to false.


**3. Nested Loops**
You can **nest loops** by placing one loop inside another. For every single iteration of the outer loop, the inner loop will run through its entire cycle from start to finish. 


**4. Loop Control Statements**
To alter the standard flow of a loop, Python provides three control statements:

1.  **`break`:** Immediately terminates the loop's execution and transfers the programme's control to the statement directly following the loop.
2.  **`continue`:** Skips the rest of the current iteration and immediately jumps back to the start of the loop to re-evaluate the condition.
3.  **`pass`:** Used as a placeholder when a statement is syntactically necessary but you do not want any code to be executed (which prevents Python from throwing an indentation error).

***Section 11 - Strings***

Following our discussion on data types and operators, let's explore **strings**—which represent text in Python—along with the powerful tools Python provides to inspect, slice, and format them.

---

**What are Strings?**

A **string** (represented by the `str` data type) is an ordered sequence of text characters. Strings are defined by surrounding text in single quotes (`'hello'`) or double quotes (`"hello"`). Double quotes are particularly useful when your string contains a single quote character, as Python will understand that the single quote is part of the text rather than the end of the string (for example, `"That is Alice's cat."`).

An important characteristic of strings is that they are **immutable sequence types**. This means that once a string is created, its characters cannot be modified, added, or removed in-place. Attempting to assign a new value to a specific index (such as `name = 'the'`) will raise a `TypeError`. If you want to modify a string, you must use slicing and concatenation to build an entirely new string by copying parts of the old one.

---

**String Indexing**
Because a string is an ordered sequence, you can access any individual character by placing its **index** inside square brackets `[]`.
*   **Positive Indexing:** Python uses 0-based indexing, meaning the first character is at index `0`, the second is at index `1`, and so on.
*   **Negative Indexing:** You can count backwards from the end of the string using negative integers, where index `-1` refers to the last character, `-2` refers to the second-to-last character, and so on.

For example, if you have a variable `spam = 'Hello, world!'`:
*   `spam` evaluates to `'H'`.
*   `spam[-1]` evaluates to `'!'`.

---

**String Slicing**
While an index retrieves a single character, a **slice** allows you to extract a subset of characters (a substring) in the form of a brand-new string. Slices are written using a start and end index separated by a colon: `[start_index:end_index]`.
*   The slice **includes the character at the start index** but **excludes the character at the end index**. For example, `spam[0:5]` evaluates to `'Hello'` (capturing characters at indexes 0, 1, 2, 3, and 4).
*   **Slicing Shortcuts:** You can leave out the index on either side of the colon. Leaving out the first index defaults to `0` (the beginning of the string), while leaving out the second index defaults to the end of the string. For example, `spam[:5]` evaluates to `'Hello'`, and `spam[7:]` evaluates to `'world!'`.
*   Slicing does not modify the original string; it simply evaluates to a new substring that you can capture in a separate variable.

---

**Useful String Methods**
A method is a function that is specifically "called on" a value. Python features several built-in string methods to analyze or transform text:
*   **Case Conversion:** `upper()` and `lower()` return a copy of the string with all letters converted to uppercase or lowercase, respectively. Because these methods return strings, you can chain other string methods directly onto their results.
*   **Case Inspection:** `isupper()` and `islower()` return `True` or `False` depending on if the string contains only uppercase or lowercase letters.
*   **Type Validation (`isX()` methods):** Methods such as `isalpha()` (letters only), `isalnum()` (letters and numbers), `isdecimal()` (numeric characters only), `isspace()` (whitespace only), and `istitle()` (title case) are highly useful for validating user input.
*   **Prefix/Suffix Check:** `startswith()` and `endswith()` verify if a string begins or ends with a specified substring.
*   **Trimming Whitespace:** `strip()`, `lstrip()`, and `rstrip()` return a new string with whitespace characters (or specific custom characters) stripped from both ends, the left end, or the right end, respectively.
*   **Justifying Text:** `ljust()`, `rjust()`, and `center()` return padded versions of the string, filling the empty space with spaces or a custom fill character to help you print neatly aligned tabular data.
*   **Joining and Splitting:**
    *   `join()` is called on a separator string and passed a list of strings, concatenating them together into a single string.
    *   `split()` does the opposite; it is called on a string and splits it (by default, wherever whitespace is found) to return a list of individual string items. You can also pass a specific delimiter string to split on (e.g., `.split('ABC')` or `.split('\n')`).
    *   `partition()` splits a string on the first occurrence of a separator and returns a 3-item tuple containing the text *before* the separator, the *separator* itself, and the text *after* the separator.

---

**String Formatting**
When you want to insert variables or other data types inside a string, doing so via standard string concatenation (using the `+` operator) is tedious and requires you to manually cast non-string values using `str()`. Python offers much cleaner approaches:

**1. String Interpolation (using `%s`)**
The `%s` operator inside a string acts as a placeholder to be replaced by the values in a tuple that follows the string. A major benefit is that Python automatically converts non-string values into their string representations behind the scenes.
*   *Example:* `'My name is %s. I am %s years old.' % (name, age)`

**2. f-strings (Formatted String Literals)**
Introduced in Python 3.6, f-strings are a simpler and more readable alternative to string interpolation. 
*   **How to write them:** You must place an **`f` prefix** directly before the starting quotation mark of the string.
*   **How they work:** You write your variables or expressions **directly inside curly braces `{}`** within the string. Python evaluates these expressions in-place and formats them automatically.
*   *Example:* `f'My name is {name}. Next year I will be {age + 1}.'`
*   *Note:* If you forget the `f` prefix, the braces and their contents will be treated as literal, unformatted text.

---

***Section 12 - List***

***List***

**A list is a value that contains multiple values in an ordered sequence.** While basic data types (like integers and strings) represent a single value, a list allows you to organize and store groups of similar or related values in a single variable. 

Lists are defined using **square brackets `[]`**, with their items separated by commas. Unlike strings, which are immutable, **lists are completely mutable**, meaning they are changeable: you can add, modify, or remove their values in-place. Furthermore, lists allow duplicate values and can even contain other lists to arrange data hierarchically.

---

**List Operations**

A **list** is a mutable data type that contains multiple values in an ordered sequence. They are created using **square brackets `[]`** with comma-delimited items. An empty list is represented simply as `[]`. 

---

**1. Creating a List**
You can define a list by assigning it to a variable. 
```python
# Creating a list of strings
fruits = ["apple", "banana", "cherry"]

# Creating an empty list
cat_names = []
```

---

**2. Accessing Items (Indexing & Slicing)**
*   **Positive Indexing:** Python lists are 0-indexed. The first item is at index `0`, the second is at index `1`, and the last index is always one less than the length of the list.
*   **Negative Indexing:** You can count backward from the end of the list using negative integers. Index `-1` refers to the last item, `-2` refers to the second-to-last item, and so on.
*   **Slicing:** You can extract a subset of items in the form of a brand-new list using the `[start:end]` slice syntax. Slicing **includes the start index but excludes the end index**. If you leave out the start index, it defaults to `0`; if you leave out the end index, it slices to the end of the list.
*   *Note:* Attempting to access an index that does not exist in the list will raise an **`IndexError`**.

```python
spam = ['cat', 'bat', 'rat', 'elephant']

print(spam)     # Output: 'cat' (Positive indexing)
print(spam[-1])    # Output: 'elephant' (Negative indexing)
print(spam[1:3])   # Output: ['bat', 'rat'] (Slicing)
print(spam[:2])    # Output: ['cat', 'bat'] (Slicing shortcut)
```

---

**3. Adding Items (Append, Extend, Insert)**
*   **`append(value)`:** Appends a new item directly to the end of the list.
*   **`extend(other_list)`:** Appends all elements from another list or iterable onto the end of the current list.
*   **`insert(index, value)`:** Inserts an item at a specified index, shifting all subsequent items down by one position.
*   *Note:* Both `append()` and `insert()` modify the list in-place and return `None` rather than the modified list.

```python
my_list =
my_list.append(4)        # my_list is now

fruits = ["apple", "banana"]
tropical = ["mango", "pineapple"]
fruits.extend(tropical)  # fruits is now ["apple", "banana", "mango", "pineapple"]

my_list.insert(1, "Hello") # my_list is now [1, "Hello", 2, 3, 4]
```
**Modifying Items**
Because lists are **mutable**, you can change an existing value at a specific index using an assignment statement.
```python
this_list = ["apple", "banana", "cherry"]
this_list.append = "blackcurrant"
# this_list is now ["apple", "blackcurrant", "cherry"]
```

---

**4. Removing Items (Remove, Pop, del)**
*   **`remove(value)`:** Searches the list and deletes the specified item. If there are duplicate items with that value, **only the first occurrence** is removed. Attempting to remove an item that is not in the list raises a `ValueError`.
*   **`pop(index)`:** Removes and returns the item at the specified index. If no index is provided, `pop()` automatically removes and returns the last item in the list.
*   **`del` Statement:** Deletes the item at a specific index. All items after the deleted index are moved up by one.

```python
spam = ['cat', 'bat', 'rat', 'elephant']

spam.remove('bat')      # spam is now ['cat', 'rat', 'elephant']
popped_item = spam.pop() # popped_item is 'elephant', spam is ['cat', 'rat']
del spam             # spam is now ['rat']
```

---

**5. Sorting & Reversing (Sort, Reverse)**
*   **`sort()`:** Sorts lists of numbers or strings alphanumerically in ascending order by default. To sort in descending order, pass the keyword argument **`reverse=True`**. 
    *   *Constraints:* You cannot sort lists containing mixed data types (such as numbers and strings combined) because Python does not know how to compare them. Also, `sort()` uses **"ASCIIbetical order"** where uppercase letters are sorted before lowercase letters.
*   **`reverse()`:** Instantly reverses the order of the items in the list in-place.
*   *Note:* Like the insertion methods, both `sort()` and `reverse()` modify the underlying list in-place and return `None`.

```python
numbers = [2, 5, 3.14, 1, -7]
numbers.sort()          # numbers is now [-7, 1, 2, 3.14, 5]

spam = ['cat', 'dog', 'moose']
spam.reverse()          # spam is now ['moose', 'dog', 'cat']
```

**6. Copying lists and Understanding References**
*   **The References Concept:** Variables do not store list values directly; they store **references** (pointers to memory addresses) where the list is located.
*   **Assigning vs. Copying:** If you assign one list variable to another (e.g., `cheese = spam`), you copy the reference. Both variables point to the exact same list. If you modify `cheese`, the list in `spam` is modified too.
*   **The `copy` Module:** To duplicate a list so you can modify it independently, you must use Python's built-in `copy` module:
    *   **`copy.copy()`** creates a duplicate, shallow copy of a mutable list.
    *   **`copy.deepcopy()`** must be used if your list contains nested lists to ensure that all inner lists are duplicated as well.

```python
import copy

spam = ['A', 'B', 'C', 'D']
cheese = copy.copy(spam)
cheese = 42

# spam remains ['A', 'B', 'C', 'D']
# cheese becomes ['A', 42, 'C', 'D']
```

````python
# 2. Accessing Items (Indexing & Slicing):

Fruit_lst = ['Banana', 120, 'Apple', 299, 'Pi', 3.141, 'Mango', 'Red', True, 'Orange', 'Red', False]

# Positive Indexing: Starts from 0 and moves from left to right.
print(Fruit_lst)       # Prints the entire list.
print(Fruit_lst[2])    # Positive indexing: Accesses the item at index 2.
print(Fruit_lst[6])    # Positive indexing: Accesses the item at index 6.

# Negative Indexing: Starts from -1 and moves from right to left.
print(Fruit_lst[-5])   # Negative indexing: Accesses the 5th item from the end.
print(Fruit_lst[-1])   # Negative indexing: Accesses the last item.


# Slicing / Range of Index

print(len(Fruit_lst))  # Returns the total number of items in the list.

# Positive Index Slicing

print("The result of 2:6 Slicing/Range is: ", Fruit_lst[2:6])      # Positive slicing: Index 2 to 5.
print("The result of 5:8 Slicing/Range is: ", Fruit_lst[5:8])      # Positive slicing: Index 5 to 7.
print("The result of 7:15 Slicing/Range is: ", Fruit_lst[7:15])    # Positive slicing: Index 7 to the end.

# Negative Index Slicing

print("The result of -9:-2 Slicing/Range is: ", Fruit_lst[-9:-2])  # Negative slicing: -9 to -3.
print("The result of -2:-9 Slicing/Range is: ", Fruit_lst[-2:-9])  # Negative indexes in reverse order; returns [].

# Mixed Positive and Negative Index Slicing

print("The result of 2:-6 Slicing/Range is: ", Fruit_lst[2:-6])    # Starts with positive index 2 and stops at negative index -6.
print("The result of -2:6 Slicing/Range is: ", Fruit_lst[-2:6])    # Starts with negative index -2 and stops at positive index 6; returns [].

# Omitting Start or Stop

print("The result of 4: Slicing/Range is: ", Fruit_lst[4:])        # Positive slicing: Index 4 to the end.
print("The result of :9 Slicing/Range is: ", Fruit_lst[:9])        # Positive slicing: Beginning to index 8.
````

***Section 13 - TUPLE***

**What is a Tuple?**

A **tuple** is an ordered sequence of values that is almost identical to a list, except for two key differences:
1.  **Syntax:** Tuples are written using round brackets—parentheses `()`—instead of lists' square brackets `[]`.
2.  **Immutability:** Unlike lists, tuples are completely **unchangeable (immutable)**. Once a tuple is created, you cannot modify, append, or remove its values. Attempting to do so will cause Python to raise a `TypeError`.

**Why use a tuple instead of a list?**
*   **Conveying Intent:** Using a tuple tells anyone reading your code that you do not intend for that specific sequence of values to change.
*   **Performance Optimizations:** Because tuples are immutable, Python can implement internal optimizations that make code running with tuples slightly faster than code running with lists.

---

**How to Create a Tuple**
You can define a tuple by placing comma-separated values inside parentheses:
```python
# Creating a standard tuple
my_tuple = (1, 2, 3, 4, 5)
eggs = ('hello', 42, 0.5)
```

**The Single-Item Tuple Gotcha**
If you want to create a tuple containing only **one value**, you must include a **trailing comma** after that value inside the parentheses. Without the comma, Python will assume you are just using parentheses for standard mathematical grouping and will evaluate it as a normal data type (like a string or integer):
```python
type(('hello',))  # Evaluates to: <class 'tuple'>
type(('hello'))   # Evaluates to: <class 'str'>
```

**Converting to a Tuple**
You can convert other sequence types (such as lists or strings) into tuples using the built-in **`tuple()`** function:
```python
my_list = ['cat', 'dog', 5]
my_tuple = tuple(my_list)  # Evaluates to: ('cat', 'dog', 5)

list_chars = list('hello')  # Evaluates to: ['h', 'e', 'l', 'l', 'o']
```

---

**Tuple Indexing & Slicing**
Because tuples are ordered, you can access their items or extract subsets using the same positive indexing, negative indexing, and slicing patterns as lists:
*   **Positive Indexing:** Accessing elements starting from index `0`.
*   **Negative Indexing:** Counting backwards from the end, where `-1` is the last item.
*   **Slicing:** Grabbing a portion of the tuple using `[start:end]` (which includes the start index but excludes the end index) to evaluate to a new tuple.

```python
thistuple = ("apple", "banana", "cherry")

print(thistuple)   # Output: "banana" (Positive indexing)
print(thistuple[-1])  # Output: "cherry" (Negative indexing)

eggs = ('hello', 42, 0.5)
print(eggs[1:3])      # Output: (42, 0.5) (Slicing)
```

---

**Tuple Packing & Unpacking**
*   **Tuple Packing** occurs when you assign multiple values into a single tuple (such as `my_tuple = (1, 2, 3)`).
*   **Tuple Unpacking** (frequently referred to in your sources as the **multiple assignment trick**) is a concise shortcut that allows you to assign the individual values inside a list or tuple to multiple variables in a single line of code.

```python
# Unpacking a list/tuple
cat = ['fat', 'gray', 'loud']
size, color, disposition = cat

print(size)         # Output: 'fat'
print(color)        # Output: 'gray'
print(disposition)  # Output: 'loud'
```
*Note: The number of variables on the left and the length of the list/tuple on the right must match **exactly**, or Python will raise a `ValueError`.*

**Function Arguments Unpacking (`*args`)**
In function definitions, you can use the **`*args`** parameter to allow a function to accept any arbitrary number of positional arguments. Behind the scenes, Python **packs** these incoming arguments into a single tuple named `args`:
```python
def add_all(*args):
    # 'args' is a tuple containing all passed arguments, e.g., (1, 2, 3)
    return sum(args)

print(add_all(1, 2, 3))  # Output: 6
```

---

**Tuple Methods**

Tuples **do not list or cover any built-in methods for tuples** (such as `.count()` or `.index()`). Instead, they focus on the fact that tuples strictly **do not support** any list methods that modify, add, or sort values—such as `.append()`, `.insert()`, `.remove()`, `.sort()`, or `.reverse()`—due to their immutable nature. 

If you ever need a mutable version of a tuple to perform such operations, you must first cast it to a list using the `list()` function, modify it, and then cast it back using `tuple()`.

***Section 14 - SET***

**What is a Set ?**

A **set** is a built-in collection used to store multiple items in a single variable. Unlike lists and dictionaries, sets represent unordered, unindexed collections that do not allow duplicate values. 

Here is everything you need to know about creating, accessing, and performing operations on sets in Python:

---

**1. Key Characteristics of a Set**
*   **Unordered & Unindexed:** Set items do not have a defined order. They can appear in a completely different order every time you use them, and you cannot refer to them by a numeric index or a key. Slicing is not supported.
*   **Unchangeable (Immutable Items):** You cannot modify the individual items inside a set after it has been created. However, you can add new items and remove existing ones.
*   **No Duplicates Allowed:** If you attempt to include duplicate values, they are automatically filtered out.
    *   *The "Truthy/Falsey" Rule:* In Python sets, **`True` and `1`** are considered the same value. Similarly, **`False` and `0`** are treated as duplicates. 

---

**2. Creating a Set**
Sets are created using curly braces **`{}`** with items separated by commas:
```python
# Creating a set
this_set = {"apple", "banana", "cherry"}

# Creating a set with duplicates (duplicates are automatically filtered)
num_set = {1, 2, 3, 3, True, 4}  # Evaluates to: {1, 2, 3, 4} because True is treated as 1
```

---

**3. Accessing Set Items**
Because sets are unindexed, you cannot retrieve individual elements using bracket notation (like `my_set`). Instead, you can:
*   Loop through the items using a **`for` loop**.
*   Check if an item is present using the **`in`** or **`not in`** membership operators.

```python
this_set = {"apple", "banana", "cherry"}

# Iterating over a set
for x in this_set:
    print(x)

# Checking membership
print("banana" in this_set)  # Output: True
```

---

**4. Adding Items**
*   **`add(item)`**: To add a single element to a set, use the `.add()` method.
*   **`update(iterable)`**: To merge items from another set (or any other iterable, like a list) into the current set, use `.update()`. This modifies the original set in-place.

```python
this_set = {"apple", "banana"}
this_set.add("orange")  # this_set is now {"apple", "banana", "orange"}

tropical = {"pineapple", "mango"}
this_set.update(tropical)  # this_set is updated to include tropical items
```

---

**5. Removing Items**
You can remove items using either `.remove()` or `.discard()`, which have an important functional difference:
*   **`remove(item)`**: Removes the specified item. If the item is not present, it **will raise an error** (causing a `KeyError` or `ValueError`).
*   **`discard(item)`**: Removes the specified item. If the item is not present, it **will NOT raise an error**.

```python
this_set = {"apple", "banana", "cherry"}

# Safe removal using discard
this_set.discard("banana")  # Removes "banana"
this_set.discard("grapes")  # Does nothing, no error is thrown!

# Removal using remove
this_set.remove("apple")    # Removes "apple"
# this_set.remove("grapes")  # This line would raise an error!
```

---

**6. Mathematical Set Operations (Joining Sets)**
One of the most powerful features of sets is their ability to perform standard mathematical set operations. Python provides both methods and operators to execute these:

**Union (`union()` or `|`)**
Combines all unique elements from both sets into a brand-new set.
```python
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}

# Using union() method
set3 = set1.union(set2)

# Using | operator
set3 = set1 | set2
```

**Intersection (`intersection()` or `&`)**
Returns a new set containing **only the duplicate elements** present in both sets.
```python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

# Using intersection() method
set3 = set1.intersection(set2)  # Evaluates to: {"apple"}

# Using & operator
set3 = set1 & set2
```

**Difference (`difference()` or `-`)**
Returns a new set containing only the elements from the **first set** that are **not present** in the second set.
```python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

# Using difference() method
set3 = set1.difference(set2)  # Evaluates to: {"banana", "cherry"}

# Using - operator
set3 = set1 - set2

# Note: Use difference_update() to modify the original set1 in-place instead
set1.difference_update(set2)  # set1 is now {"banana", "cherry"}
```

**Symmetric Difference (`symmetric_difference()` or `^`)**
Returns a new set containing all elements **except** the ones present in both sets (keeps everything except the duplicates).
```python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

# Using symmetric_difference() method
set3 = set1.symmetric_difference(set2)  # Evaluates to: {"banana", "cherry", "google", "microsoft"}

# Using ^ operator
set3 = set1 ^ set2
```

---

**7. Set Comprehension**
Although your sources do not explicitly demonstrate "Set Comprehension" syntax like they do for lists and dictionaries, sets are sequence collection types that can be created dynamically. If you would like to generate a set dynamically using a single-line comprehension, it uses curly braces `{}` with the same logical structure as list comprehensions:

```python
* Extracting unique squared numbers for even values
even_squares_set = {x * x for x in range(10) if x % 2 == 0}
* Evaluates to the set: {0, 16, 4, 36, 64}
```

***Section 14 - DICTIONARY***

**What is a dictionary?**

**A dictionary is a mutable collection of many values.** While lists use ordered integer indexes, dictionaries use index values of various data types (such as strings, integers, floats, or even tuples). These unique index values are called **keys**, and a key combined with its associated value is known as a **key-value pair**.

In code, dictionaries are written using curly braces **`{}`** with colons separating keys and values, and commas separating each pair. An empty dictionary is represented simply as `{}`.

---

**1. Dictionaries vs. Lists**
*   **Unordered:** Unlike lists, the items inside a dictionary are completely **unordered**. Because there is no concept of a "first" item, two dictionaries containing the exact same key-value pairs are considered equal, even if the pairs are typed in a different order.
*   **No Slicing:** Since dictionaries are unordered, they **cannot be sliced** like lists or strings.
*   **KeyErrors:** Attempting to access a key that does not exist in a dictionary will raise a **`KeyError`** (similar to a list's "out-of-range" `IndexError`).
*   **Insertion Order (Python 3.7+):** In Python 3.7 and later, dictionaries will remember the insertion order of their key-value pairs if you convert them into a sequence (like a list). However, they remain unordered in the sense that you still cannot access their items using numerical indexes (such as `my_dict`) unless `0` is an explicit key.

---

**2. Basic Dictionary Operations**

**Creating and Accessing Items**
You can access a value by placing its key inside square brackets `[]`:
```python
my_cat = {'size': 'fat', 'color': 'gray', 'disposition': 'loud'}
print(my_cat['size'])  # Output: 'fat'
```

**Adding or Modifying Key-Value Pairs**
To add a new key-value pair or update an existing one, use the assignment operator (`=`) with the key name in square brackets:
```python
my_cat['color'] = 'black'  # Modifies existing key 'color'
my_cat['age'] = 5          # Adds a brand-new key 'age'
```

**Deleting Key-Value Pairs**
*   **`del` Statement:** Deletes a key-value pair using the `del` keyword:
    ```python
    del my_cat['disposition']
    ```
*   **`pop()` Method:** Removes the specified key and returns its associated value:
    ```python
    age = my_cat.pop('age')  # Removes 'age' and stores 5 in variables
    ```
*   **`popitem()` Method:** Removes and returns the last inserted key-value pair.

---

**3. Essential Dictionary Methods**

**Keys, Values, and Items**
Python provides three list-like methods to easily extract and iterate over the contents of a dictionary:
1.  **`keys()`**: Returns a list-like view of all keys in the dictionary.
2.  **`values()`**: Returns a list-like view of all values.
3.  **`items()`**: Returns a list-like view of the key-value pairs as a sequence of **tuples**.

You can loop through these views directly or convert them into true lists using the `list()` function:
```python
spam = {'color': 'red', 'age': 42}

# Looping through keys and values simultaneously
for k, v in spam.items():
    print(f"Key: {k}, Value: {v}")
```

**Checking for Existence**
You can use the **`in`** and **`not in`** operators to check if a specific key or value exists inside a dictionary. By default, checking directly on the dictionary checks against its **keys**:
```python
print('name' in spam)         # Checks keys (evaluates to False)
print('red' in spam.values()) # Checks values (evaluates to True)
```

**The `get()` Method**
To safely retrieve a value without risking a `KeyError`, use the **`get()`** method. It takes two arguments: the key to look up and a **fallback value** to return if the key does not exist:
```python
picnic_items = {'apples': 5, 'cups': 2}
print(picnic_items.get('apples', 0)) # Output: 5
print(picnic_items.get('eggs', 0))   # Output: 0 (since 'eggs' is missing)
```

**The `setdefault()` Method**
You will often want to set a default value for a key **only if** that key does not already exist. The **`setdefault()`** method lets you do this in a single line:
```python
spam = {'name': 'Pooka', 'age': 5}
spam.setdefault('color', 'black') # Sets 'color' to 'black' because it didn't exist
spam.setdefault('age', 10)        # Does nothing because 'age' already exists (returns 5)
```

**The `update()` Method**
Merges another dictionary (or an iterable of key-value pairs) directly into the current dictionary, overwriting any matching keys:
```python
spam.update({'age': 6, 'species': 'cat'})
```

---

**4. Pretty Printing (`pprint` module)**
If you are working with large or nested dictionaries, printing them with the standard `print()` function can be messy and hard to read. You can import the **`pprint`** module to "pretty print" your dictionary, which automatically cleans up the formatting and sorts the keys alphabetically:
```python
import pprint
message = 'It was a bright cold day in April, and the clocks were striking thirteen.'
count = {}

for character in message:
    count.setdefault(character, 0)
    count[character] += 1

pprint.pprint(count)
```

````python
# Q — Dictionary Keys and Values

data = {
    1: "Python",
    "1": "Java",
    True: "C++"
}

# 1. Check whether 1 exists as a key.
# 2. Check whether True exists as a key.
# 3. Check whether "1" exists as a key.
# 4. Check whether "Python" exists as a value.

print("Dose 1 exist as a key in data: ", 1 in data)
print("Does True exist as a key in data: ", True in data)
print("Dose '1' exist as a key in data: ", "1" in data)
print("Does 'Python' exist as a value in data: ", "Python" in data.values())
````

````python
# Q — Checking Keys vs. Values

info = {
    "name": "Alex",
    "language": "Python",
    "level": "Advanced"
}

# 1. Check whether "Python" exists as a key.
# 2. Check whether "Python" exists as a value.
# 3. Check whether "python" does NOT exist as a value.
# 4. Check whether "language" exists as a value.

print("Does 'Python' exist as a key in info: ", "Python" in info)
print("Does 'Python' exist as a value in info: ", "Python" in info.values())
print ("'python' does NOT exist as a value in info: ", "python" not in info.values())
print("Does 'language' exist as a value in info: ", "language" in info.values())
````

````python
# Q — Duplicate Values

d = {
    "a": 10,
    "b": 20,
    "c": 10
}

# 1. Check whether 10 exists as a key.
# 2. Check whether 10 exists as a value.
# 3. Check whether "a" exists as a value.
# 4. Check whether "a" exists as a key.
# 5. Check whether 30 does NOT exist as a value.

print("Does 10 exist as a key in d: ", 10 in d)
print("Does 10 exist as a value in d: ", 10 in d.values())
print("Does 'a' exist as a value in d", "a" in d.values())
print("Does 'a' exist as a key in d: ", "a" in d)
print("30 Does NOT exist as a value in d: ", 30 not in d.values())
````

````python
# Write a program to get details from n employees and store them in the dict.

Employee_Details = dict()

count = int(input("Enter How Many Employees' Data You Want to Enter: "))
print('\n')

while True:
    Employee_ID = input("Enter Your Employee ID: ")

    if Employee_ID in Employee_Details:
        print("Employee ID Already Exists")
        print('\n')
        continue

    else:
        Employee_Data = {}

        employee_id = Employee_ID
        name = input("Enter Employee Name: ")
        salary = float(input("Enter Employee Salary: "))
        position = input("Enter Employee Position: ")
        date_of_join = input("Enter Employee Date of Joining: ")
        department = input("Enter Employee Department Name: ")

    Employee_Data['Employee_ID'] = employee_id
    Employee_Data['Name'] = name
    Employee_Data['Salary'] = salary
    Employee_Data['Position'] = position
    Employee_Data['Date_of_Join'] = date_of_join
    Employee_Data['Department'] = department

    Employee_Details[Employee_ID] = Employee_Data

    count -= 1

    if count == 0:
        print("Program Completed Successfully")
        break

import pprint
pprint.pprint(Employee_Details, sort_dicts=False)
````
